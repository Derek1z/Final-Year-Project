# 01. Exploratory Data Analysis for PMSM Fault Telemetry

This notebook analyzes electrical telemetry signals ($I_d, I_q, V_d, V_q, W_m$) collected from Permanent Magnet Synchronous Motor (PMSM) simulations under various fault operating conditions.

```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Add project root to sys.path
repo_root = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
if str(repo_root) not in sys.path:
    sys.path.append(str(repo_root))

from src.data_loader import load_and_preprocess_data
```

## 1. Load Dataset

```python
data_path = repo_root / 'data' / 'Classification.csv'
df = pd.read_csv(data_path)
clean_df = df.loc[:, ~df.columns.str.contains('^Unnamed')]
print(f"Dataset Shape: {clean_df.shape}")
clean_df.head()
```

## 2. Target Class Distribution

```python
plt.figure(figsize=(8, 5))
sns.countplot(x='Output', data=clean_df, palette='viridis')
plt.title('PMSM Fault Class Counts')
plt.xlabel('Fault Class (0=Healthy, 1=Phase A, 2=Phase A&B, 3=Short Circuit, 4=Open Circuit)')
plt.ylabel('Sample Count')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 3. Feature Distributions & Boxplots

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=clean_df.drop(columns=['Output']))
plt.title('Feature Distributions (Id, Iq, Vd, Vq, Wm)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 4. Correlation Heatmap

```python
plt.figure(figsize=(10, 8))
sns.heatmap(clean_df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('PMSM Telemetry Feature Correlation Matrix')
plt.show()
```
